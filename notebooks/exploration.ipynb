# Dataset Exploration
Quick look at the raw dataset before running `src/data_pipeline.py`.

```python
from datasets import load_dataset
import yaml

with open('../configs/training_config.yaml') as f:
    config = yaml.safe_load(f)

ds = load_dataset(config['dataset']['name_or_path'], split='train')
print(f"Total examples: {len(ds)}")
ds[0]
```

```python
# Inspect field lengths to sanity-check max_seq_length before tokenizing the full dataset
import numpy as np

lengths = [len((ex.get('instruction','') + ex.get('context','') + ex.get('response','')).split()) for ex in ds.select(range(500))]
print(f"mean words: {np.mean(lengths):.1f}, p95: {np.percentile(lengths, 95):.1f}, max: {np.max(lengths)}")
```

```python
# Preview a formatted example using the model's own chat template
import sys
sys.path.append('../src')
from transformers import AutoTokenizer
from data_pipeline import format_chatml

tokenizer = AutoTokenizer.from_pretrained(config['model']['base_model_id'], trust_remote_code=True)
formatted = format_chatml(ds[0], tokenizer, config['dataset']['system_prompt'])
print('--- prompt (no response) ---')
print(formatted['prompt'])
print('--- full text (incl. response) ---')
print(formatted['text'])
```
